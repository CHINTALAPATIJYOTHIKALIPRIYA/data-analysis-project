```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/content/SampleSuperstore.csv')
```

```python
category_sales = df.groupby('Category')['Sales'].sum()

print(category_sales)
```

```python
category_sales.plot(kind='bar')

plt.title('Sales by Category')
plt.xlabel('Category')
plt.ylabel('Sales')

plt.show()
```

```python
region_profit = df.groupby('Region')['Profit'].sum()

print(region_profit)
```

```python
region_profit.plot(kind='bar')

plt.title('Profit by Region')
plt.xlabel('Region')
plt.ylabel('Profit')

plt.show()
```

```python
loss_subcategories = df.groupby('Sub-Category')['Profit'].sum()

loss_subcategories = loss_subcategories[loss_subcategories < 0]

print(loss_subcategories)
```

```python
discount_profit = df.groupby('Discount')['Profit'].sum()

print(discount_profit)
```

# AI Business Insights

- Technology category generated the highest sales among all categories.

- West region showed stronger profitability compared to other regions.

- High discounts negatively impacted overall profit.

- Some subcategories showed consistent losses and may require pricing optimization.

- Business can improve profitability by reducing excessive discount strategies.
